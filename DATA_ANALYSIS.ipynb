```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 5)
```

```python
# Load Dataset
df = pd.read_csv("/content/drive/MyDrive/DataCoSupplyChainDataset.csv.zip", compression='zip', encoding='latin1')
```

```python
plt.figure()
sns.countplot(data=df, x='Delivery Status', palette='Set2')
plt.title('Delivery Status Distribution across Orders', fontsize=14, fontweight='bold')
plt.xlabel('Delivery Status', fontsize=12)
plt.ylabel('Order Count', fontsize=12)
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
# LATE DELIVERY RISK BY SHIPPING MODE
plt.figure()
sns.barplot(data=df, x='Shipping Mode', y='Late_delivery_risk', ci=None, palette='viridis')
plt.title('Late Delivery Risk Proportion by Shipping Mode', fontsize=14, fontweight='bold')
plt.xlabel('Shipping Mode', fontsize=12)
plt.ylabel('Late Delivery Rate', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
#  TOP 10 REGIONS BY TOTAL SALES
top_regions = df.groupby('Order Region')['Sales'].sum().reset_index().sort_values(by='Sales', ascending=False).head(10)

plt.figure()
sns.barplot(data=top_regions, y='Order Region', x='Sales', palette='Blues_r')
plt.title('Top 10 Order Regions by Revenue Generation', fontsize=14, fontweight='bold')
plt.xlabel('Total Sales ($)', fontsize=12)
plt.ylabel('Order Region', fontsize=12)
plt.tight_layout()
plt.show()
```
